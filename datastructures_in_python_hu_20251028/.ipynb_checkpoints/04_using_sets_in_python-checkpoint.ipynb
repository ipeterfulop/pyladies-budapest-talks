# Set (halmaz) használata Pythonban
A **set** (halmaz) egy **rendezetlen** gyűjtemény, amely **csak egyedi** elemeket tárolhat. Nincs indexelése, és elemei **immutable** objektumok kell legyenek (pl. lista nem lehet set eleme, de tuple igen). A halmazok fő előnye a **rendkívül gyors ellenőrzése annak, hogy adott elem szerepel-e benne vagy sem**.

## Set inicializálása

### Inicializálás `{}` kapcsos zárójelek használatával

````python
# Figyelem! Az üres {} egy üres SZÓTÁRAT hoz létre, NEM üres halmazt!
ures_szotar = {}
print(type(ures_szotar))  # <class 'dict'>

# Üres halmazhoz a set() konstruktort kell használni (lásd következő szekció)
````

````python
# halmaz inicializálása string elemekkel
protokollok = {'HTTP', 'HTTPS', 'FTP', 'SSH', 'DNS'}
print(type(protokollok))
print(f"Protokollok: {protokollok}")
````

````python
# halmaz inicializálása int elemekkel
nyitott_portok = {80, 443, 22, 21, 8080}
print(f"Nyitott portok: {nyitott_portok}")

# duplikált elemek automatikusan eltávolításra kerülnek
portok_duplikalttal = {80, 443, 22, 80, 443, 21}
print(f"Duplikált portok halmaza: {portok_duplikalttal}")
print(f"Elemek száma: {len(portok_duplikalttal)}")
````

````python
# a halmaz vegyes (de immutable) típusú elemeket is tartalmazhat
vegyes_halmaz = {80, 'SSH', 3.14, True, (1, 2, 3)}
for elem in vegyes_halmaz:
    print(f"<{elem}> típusa {type(elem)}")

# lista NEM lehet halmaz eleme (mert mutable)
try:
    hibas_halmaz = {[1, 2], 'teszt'}
except TypeError as e:
    print(f"\nHiba: {e} - lista nem lehet halmaz eleme!")
````

### Inicializálás `set()` konstruktor használatával

````python
# üres halmaz inicializálása
ures_halmaz = set()
print(type(ures_halmaz))
print(ures_halmaz)

# halmaz létrehozása listából (duplikátumok automatikusan kiszűrődnek)
port_lista = [80, 443, 22, 80, 21, 443]
egyedi_portok = set(port_lista)
print(f"Listából létrehozott halmaz: {egyedi_portok}")

# halmaz létrehozása stringből (karakterenként)
betuk = set('abrakadabra')
print(f"Stringből létrehozott halmaz: {betuk}")

# halmaz létrehozása tuple-ből
komponensek_tuple = ('RAM', 'CPU', 'SSD', 'RAM', 'GPU')
komponensek_halmaz = set(komponensek_tuple)
print(f"Tuple-ből létrehozott halmaz: {komponensek_halmaz}")

# halmaz létrehozása szótár kulcsaiból
pc_konfiguracio = {'CPU': 'i9', 'RAM': 64, 'SSD': '1TB'}
kulcsok_halmaz = set(pc_konfiguracio)
print(f"Szótár kulcsaiból létrehozott halmaz: {kulcsok_halmaz}")
````

## Alapműveletek halmazokkal

### Elem hozzáadása és törlése

````python
telepitett_szoftverek = {'VS Code', 'Firefox', 'Git', 'Python'}
print(f"Telepített szoftverek: {telepitett_szoftverek}")

# elem hozzáadása (add)
telepitett_szoftverek.add('Docker')
print(f"Docker hozzáadva: {telepitett_szoftverek}")

# már létező elem hozzáadása nem okoz hibát, de nincs hatása
telepitett_szoftverek.add('Git')
print(f"Git újra hozzáadva (nincs változás): {telepitett_szoftverek}")

# több elem hozzáadása egyszerre (update)
telepitett_szoftverek.update(['Slack', 'Postman', 'Node.js'])
print(f"Több elem hozzáadva (update): {telepitett_szoftverek}")
````

````python
telepitett_szoftverek = {'VS Code', 'Firefox', 'Git', 'Python', 'Docker', 'Slack'}
print(f"Kiindulás: {telepitett_szoftverek}")

# elem eltávolítása (remove) - KeyError-t dob, ha az elem nem létezik
telepitett_szoftverek.remove('Slack')
print(f"Slack eltávolítva (remove): {telepitett_szoftverek}")

# elem eltávolítása (discard) - NEM dob hibát, ha az elem nem létezik
telepitett_szoftverek.discard('Notepad')  # nem létezik, de nem okoz hibát
telepitett_szoftverek.discard('Docker')
print(f"Docker eltávolítva (discard): {telepitett_szoftverek}")

# pop() - egy véletlenszerű elemet emel ki (a halmaz rendezetlen!)
kiemelt = telepitett_szoftverek.pop()
print(f"Kiemelt elem: {kiemelt}, maradék: {telepitett_szoftverek}")

# clear() - minden elem törlése
telepitett_szoftverek.clear()
print(f"Törlés után: {telepitett_szoftverek}")
````

### Tartalmazás vizsgálata (`in`)
A halmazok egyik legnagyobb előnye, hogy az `in` operátor rendkívül gyors (átlagosan O(1) időkomplexitás), szemben a listával (O(n)).

````python
engedélyezett_portok = {80, 443, 22, 8080, 3306}

# elem tartalmazásának vizsgálata
vizsgalt_port = 443
print(f"A {vizsgalt_port}-es port engedélyezett? {vizsgalt_port in engedélyezett_portok}")

vizsgalt_port = 25
print(f"A {vizsgalt_port}-ös port engedélyezett? {vizsgalt_port in engedélyezett_portok}")

# nem tartalmazás vizsgálata
print(f"A 25-ös port NINCS engedélyezve? {25 not in engedélyezett_portok}")
````

````python
# sebesség összehasonlítás: halmaz vs lista
import time

N = 1_000_000
nagy_lista = list(range(N))
nagy_halmaz = set(range(N))
keresett_elem = N - 1  # az utolsó elem (legrosszabb eset a listánál)

# keresés listában
start = time.time()
for _ in range(100):
    keresett_elem in nagy_lista
lista_ido = time.time() - start

# keresés halmazban
start = time.time()
for _ in range(100):
    keresett_elem in nagy_halmaz
halmaz_ido = time.time() - start

print(f"Lista keresés ({N:_} elem, 100x): {lista_ido:.4f} mp")
print(f"Halmaz keresés ({N:_} elem, 100x): {halmaz_ido:.6f} mp")
print(f"A halmaz ~{lista_ido / halmaz_ido:.0f}x gyorsabb!")
````

### Bejárás (iterálás)

````python
operacios_rendszerek = {'Windows', 'macOS', 'Linux', 'ChromeOS'}

# elem-alapú iteráció (a sorrend nem garantált!)
print("Elem-alapú iteráció:")
for os_nev in operacios_rendszerek:
    print(f"  OS: {os_nev}")

# enumerate() használata (ritkábban használt halmazokkal, mert nincs fix sorrend)
print("\nEnumerate-tel:")
for index, os_nev in enumerate(operacios_rendszerek):
    print(f"  {index}. {os_nev}")
````

## Halmazműveletek
A Python halmazok támogatják a matematikai halmazműveleteket, amelyeket mind metódusokkal, mind operátorokkal elvégezhetünk.

### Unió (union) `|`
Két halmaz **összes** elemét tartalmazza (duplikátumok nélkül).

````python
frontend_technologiak = {'HTML', 'CSS', 'JavaScript', 'React'}
backend_technologiak = {'Python', 'Node.js', 'JavaScript', 'SQL'}

# unió operátorral
osszes_technologia = frontend_technologiak | backend_technologiak
print(f"Összes technológia (operátor): {osszes_technologia}")

# unió metódussal
osszes_technologia_2 = frontend_technologiak.union(backend_technologiak)
print(f"Összes technológia (metódus): {osszes_technologia_2}")
````

### Metszet (intersection) `&`
Csak azokat az elemeket tartalmazza, amelyek **mindkét** halmazban szerepelnek.

````python
frontend_technologiak = {'HTML', 'CSS', 'JavaScript', 'React'}
backend_technologiak = {'Python', 'Node.js', 'JavaScript', 'SQL'}

# metszet operátorral
kozos_technologiak = frontend_technologiak & backend_technologiak
print(f"Közös technológiák (operátor): {kozos_technologiak}")

# metszet metódussal
kozos_technologiak_2 = frontend_technologiak.intersection(backend_technologiak)
print(f"Közös technológiák (metódus): {kozos_technologiak_2}")
````

### Különbség (difference) `-`
Azokat az elemeket tartalmazza, amelyek az **első** halmazban megvannak, de a **másodikban** nem.

````python
frontend_technologiak = {'HTML', 'CSS', 'JavaScript', 'React'}
backend_technologiak = {'Python', 'Node.js', 'JavaScript', 'SQL'}

# különbség operátorral
csak_frontend = frontend_technologiak - backend_technologiak
print(f"Csak frontend: {csak_frontend}")

csak_backend = backend_technologiak - frontend_technologiak
print(f"Csak backend: {csak_backend}")

# különbség metódussal
csak_frontend_2 = frontend_technologiak.difference(backend_technologiak)
print(f"Csak frontend (metódus): {csak_frontend_2}")
````

### Szimmetrikus különbség (symmetric difference) `^`
Azokat az elemeket tartalmazza, amelyek **pontosan az egyik** halmazban szerepelnek, de **nem mindkettőben**.

````python
frontend_technologiak = {'HTML', 'CSS', 'JavaScript', 'React'}
backend_technologiak = {'Python', 'Node.js', 'JavaScript', 'SQL'}

# szimmetrikus különbség operátorral
nem_kozos = frontend_technologiak ^ backend_technologiak
print(f"Csak az egyikben (operátor): {nem_kozos}")

# szimmetrikus különbség metódussal
nem_kozos_2 = frontend_technologiak.symmetric_difference(backend_technologiak)
print(f"Csak az egyikben (metódus): {nem_kozos_2}")
````

### Részhalmaz és szuperhalmaz vizsgálat

````python
osszes_port = {21, 22, 25, 80, 443, 3306, 5432, 8080}
web_portok = {80, 443, 8080}
adatbazis_portok = {3306, 5432}

# részhalmaz vizsgálat (issubset / <=)
print(f"A web portok részhalmaza az összes portnak? {web_portok.issubset(osszes_port)}")
print(f"Operátorral: {web_portok <= osszes_port}")

# szuperhalmaz vizsgálat (issuperset / >=)
print(f"\nAz összes port szuperhalmaza a web portoknak? {osszes_port.issuperset(web_portok)}")
print(f"Operátorral: {osszes_port >= web_portok}")

# diszjunkt halmazok vizsgálata (isdisjoint - nincs közös elem)
print(f"\nA web és adatbázis portoknak van közös eleme? {not web_portok.isdisjoint(adatbazis_portok)}")
print(f"A web és adatbázis portok diszjunktak? {web_portok.isdisjoint(adatbazis_portok)}")
````

## Halmazműveletek helyben (in-place)
Az eddigi műveletek új halmazt hoztak létre. A helyben módosító változatok az eredeti halmazt frissítik.

````python
aktiv_felhasznalok = {'anna', 'bela', 'cecil'}
uj_bejelentkezok = {'cecil', 'diana', 'edit'}

print(f"Aktív felhasználók (előtte): {aktiv_felhasznalok}, memória cím: {id(aktiv_felhasznalok)}")

# |= (update): helyben unió
aktiv_felhasznalok |= uj_bejelentkezok
print(f"Aktív felhasználók (unió után): {aktiv_felhasznalok}, memória cím: {id(aktiv_felhasznalok)}")

# &= (intersection_update): helyben metszet
engedélyezett = {'anna', 'diana', 'edit', 'feri'}
aktiv_felhasznalok &= engedélyezett
print(f"Engedélyezett aktív felhasználók: {aktiv_felhasznalok}")

# -= (difference_update): helyben különbség
kileptettek = {'edit'}
aktiv_felhasznalok -= kileptettek
print(f"Kiléptetés után: {aktiv_felhasznalok}")
````

## Set comprehensions (halmaz-értelmezés)
A halmaz-értelmezés szintaxisa megegyezik a lista-értelmezéssel, de `{}` kapcsos zárójeleket használunk.

````python
# alap halmaz-értelmezés: fájlkiterjesztések kinyerése
fajlok = ['main.py', 'index.html', 'style.css', 'app.py', 'readme.md', 'config.py']

kiterjesztesek = {fajl.split('.')[-1] for fajl in fajlok}
print(f"Egyedi fájlkiterjesztések: {kiterjesztesek}")
````

````python
# szűréssel: csak a rendszerportokat gyűjtjük ki
portok = [80, 443, 22, 8080, 3306, 21, 80, 443]

rendszer_portok = {p for p in portok if p < 1024}
print(f"Egyedi rendszerportok: {rendszer_portok}")
````

````python
# transzformációval: IP-címek utolsó oktetjének kinyerése
ip_cimek = ['192.168.1.10', '192.168.1.20', '10.0.0.5', '192.168.1.10', '10.0.0.5']

egyedi_halozatok = {ip.rsplit('.', 1)[0] for ip in ip_cimek}
print(f"Egyedi hálózati címek: {egyedi_halozatok}")
````

## Frozenset (módosíthatatlan halmaz)
A `frozenset` a set immutable változata. Nem lehet elemeket hozzáadni vagy eltávolítani belőle, viszont szótárkulcsként vagy más halmaz elemeként is használható.

````python
# frozenset létrehozása
alapveto_portok = frozenset({80, 443, 22})
print(f"Frozenset: {alapveto_portok}, típusa: {type(alapveto_portok)}")

# a halmazműveletek működnek (új frozenset-et adnak vissza)
tovabbi_portok = frozenset({443, 8080, 3306})
print(f"Unió: {alapveto_portok | tovabbi_portok}")
print(f"Metszet: {alapveto_portok & tovabbi_portok}")

# módosítás NEM lehetséges
try:
    alapveto_portok.add(21)
except AttributeError as e:
    print(f"\nHiba: {e}")
````

````python
# frozenset használata szótárkulcsként
# Pl. hozzáférési szabályok: portok kombinációja -> engedélyezett-e
hozzaferesi_szabalyok = {
    frozenset({80, 443}): 'Web forgalom engedélyezve',
    frozenset({22}): 'SSH hozzáférés engedélyezve',
    frozenset({3306, 5432}): 'Adatbázis hozzáférés engedélyezve'
}

keresett = frozenset({80, 443})
print(f"Szabály: {hozzaferesi_szabalyok[keresett]}")
````

## Mikor ...?

### Mikor ajánlott a set használata?

- **egyediség garantálása**: ha biztosítani akarjuk, hogy egy gyűjtemény ne tartalmazzon duplikált elemeket. Példa: egyedi felhasználónevek, egyedi IP-címek.
- **gyors tartalmazás-vizsgálat**: ha gyakran kell ellenőrizni, hogy egy elem része-e a gyűjteménynek. A halmaz O(1) idő alatt válaszol, míg a lista O(n) időt igényel.
- **halmazműveletek szükségesek**: ha uniót, metszetet, különbséget vagy szimmetrikus különbséget kell számítani (pl. közös jogosultságok, hiányzó konfigurációk).
- **duplikátumok eltávolítása listából**: a `set(lista)` a legegyszerűbb módja az egyedi elemek kiszűrésének.

### Mikor NEM ajánlott a set használata?

- **sorrend fontos**: a halmaz rendezetlen, tehát ha az elemek sorrendje számít, használjunk listát vagy tuple-t.
- **indexelés szükséges**: a halmaz nem támogatja az indexelést (`halmaz[0]` hibát dob).
- **mutable elemeket kell tárolni**: a halmaz csak immutable elemeket fogad el (pl. lista nem lehet halmaz eleme).
- **kulcs-érték párok kellenek**: ha az elemekhez értékeket is szeretnénk rendelni, használjunk szótárat.
