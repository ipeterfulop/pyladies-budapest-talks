## 1.1. Szemantika: paraméterek és argumentumok
Tisztázzuk a paraméterek és argumentuomok közötti jelentésbeli különbséget

````python
# a és b a függvény PARAMÉTEREI — a függvény definíciójában szereplő változónevek
def osszead(a, b, c):
    return a + b + c


x = 4
y = -1
# x és y a függvény ARGUMENTUMAI — a híváskor ténylegesen átadott értékek(referenciák)
# átadás: pozíció szerinti 
eredmeny = osszead(x, y, 0)
print("eredmeny: ", eredmeny)
eredmeny = osszead(y, x, 0)
print(f"eredmeny: {eredmeny}")

t = 2
u = 3
# u és t a függvény ARGUMENTUMAI
# névvel ellátott (keyword) argumentumok — a paraméter nevével adjuk át az értékeket(referenciákat)
eredmeny = osszead(a=t, b=u, c=3)
print(eredmeny)
eredmeny = osszead(b=t, c=1, a=u)
````

## 1.2. Már korábban is használtunk függvényeket

A lista elemeinek összegzésére az előző előadáson már használtuk a `sum()` függvényt. Sőt, a képernyőre való kiíratáshoz a `print()` függvényt már a Pythonnal való ismerkedés első lépései során is használtuk.

Ezek a Python beépített függvényei.

````python
# példa a sum beépített függvény használatára
l1 = [ 2, 8, 9, -1, 4]
osszeg = sum(l1)
print(f" Az <l1> lista tartalma {l1}, elemeinek összege: {osszeg}")

# példa a print beépített függvény használatára
print("Jó estét Pyladies Budapest!")

# beépített függvények típusának felderítése
print(type(sum))

help(sum)
````

````python
# 3. haladóbb megközelítés: az inspect modullal kiszűrjük csak a függvényeket, és kiírjuk a dokumentációjukat
import builtins
import inspect

for name in dir(builtins):
    obj = getattr(builtins, name)
    if inspect.isbuiltin(obj):
        print("\n\n", name, "\n=========\n", obj.__doc__)
        # szándékosan nem használtam f-stringet
````
