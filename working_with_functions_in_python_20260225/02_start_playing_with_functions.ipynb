## 2.1. Saját függvények létrehozása

### 2.1.1 `koszonj_neki()` – Az első saját függvényed

Kezdjük valami egyszerűvel: írjunk egy függvényt, ami **köszön** valakinek. Két paramétert kap:
- `o` – az üdvözölt személy neve
- `en` – a köszönő neve

Ez a függvény nem ad vissza semmit — csak kiír egy üzenetet a `print()` segítségével.

````python
def koszonj_neki(o, en):
    """Üdvözli a megadott személyt a köszönő nevében."""
    print(f"Szia, {o}! Én {en} vagyok, örülök, hogy itt vagy!")
````

````python
# Próbáld ki!
koszonj_neki("Eszter", "Kata")
koszonj_neki(o="Bence", en="Nóri")

# nevesített argumentumoknál a sorrend nem számít
koszonj_neki(en="Péter", o="Kata")

print(f"A függvény leírása: {koszonj_neki.__doc__}")
````

**Mit történt itt?**

1. A `def` kulcsszóval létrehoztunk egy új függvényt `koszonj_neki` néven.
2. A zárójelben megadtuk a **paraméterek** (`o` és `en`) szerinti **argumentumokat**.
3. A függvény kódját (function body) **behúzással** (indentálással) indítottuk — minden sor, ami beljebb van, a függvényhez tartozik.
4. Amikor meghívtuk (`koszonj_neki("Eszter", "Kata")`), az `o` helyére `"Eszter"` került, az `en` helyére `"Kata"`.

A hármas idézőjeles szöveg a függvény elején a **docstring** — ez egy rövid leírás arról, mit csinál a függvény. Nem kötelező, de nagyon hasznos!

### 2.1.2 `mond_meg_hogy_paros_e()` – Döntés a függvényen belül

Most írjunk egy függvényt, ami megmondja, hogy egy szám **páros** vagy **páratlan**. Ez is csak `print()`-tel ír ki — egyelőre még nem használunk `return`-t.

````python
def mond_meg_hogy_paros_e(szam):
    """Kiírja, hogy a szám páros-e vagy páratlan."""
    if szam % 2 == 0:
        print(f"A(z) {szam} páros szám.")
    else:
        print(f"A(z) {szam} páratlan szám.")
````

````python
mond_meg_hogy_paros_e(42)
mond_meg_hogy_paros_e(szam=17)
mond_meg_hogy_paros_e(int("2"))
````

Mi történik, ha a megállapítás eredményét egy változóba próbáljuk menteni:

````python
eredmeny = mond_meg_hogy_paros_e(42)
print(f"Az eredmény értéke: {eredmeny}")
print(f"Az eredmény típusa: {type(eredmeny)}")
````

**Megjegyzés:** Ha egy függvényben nincs `return` utasítás, a Python automatikusan `None`-t ad vissza. A `print()` csak **kiír** a képernyőre, de nem **térít vissza** értéket.

### 1.3 `paros_e()` – Visszatérési érték és típusjelölés

Ismételjük meg a fenti logikát, de egy módosítással: a függvény **visszaad** egy értéket (`True` vagy `False`) ahelyett, hogy kiírná.

Két új dolgot is bevezetünk:
- a **`return`** utasítást — ezzel adja vissza a függvény az eredményt, ezt hívjuk **visszatérési értéknek** (*return value*)
- a **típusjelölést** (*type hint*) - nem kötelező — ami jelzi, milyen típusú paramétert várunk és mit adunk vissza

````python
def paros_e(szam: int) -> bool:
    """Visszaadja, hogy a szám páros-e."""
    return szam % 2 == 0
````

````python
# Most már el tudjuk menteni az eredményt!
eredmeny = paros_e(42)
print(f"42 páros? {eredmeny}")
print(f"Az eredmény típusa: {type(eredmeny)}")

eredmeny1:bool = paros_e(21)
print(f"eredmeny1: {eredmeny1}")
````

````python
# A függvényt használhatjuk feltételben is:
if paros_e(7):
    print("A 7 páros.")
else:
    print("A 7 páratlan.")
````

**Mi a különbség a két megközelítés között?**

| | `mond_meg_hogy_paros_e(42)` | `paros_e(42)` |
|---|---|---|
| **Mit csinál?** | Kiír a képernyőre | Visszaad egy bool típusú értéket |
| **Visszatérési érték** | `None` | `True` |
| **Felhasználható logikai feltételben?** | Nem | Igen! |

A típusjelölés (`szam: int` és `-> bool`) azt jelzi, hogy a függvény egész számot vár, és logikai értéket (`True`/`False`) ad vissza. Ez nagyban segíti a kód olvashatóságát — viszont fontos tudni, hogy a Python ezt **nem kényszeríti ki**: ha más típusú értékkel hívod meg a függvényt, nem kapsz hibaüzenetet. A típusjelölés inkább egy „útjelző tábla" neked és a kódszerkesztődnek, nem egy szigorú szabály.

**Tanulság:** A `return` utasítással a függvény **visszaad** egy értéket a hívónak. Ez sokkal rugalmasabb, mint a `print()`, mert az eredményt változóba mentheted, feltételben használhatod, vagy továbbadhatod egy másik függvénynek.

### 1.4 `szamold_meg_a_maganhangzokat()` – String bejárás és számláló

Írjunk egy kicsit összetettebb függvényt: megszámolja, hány **magánhangzó** van egy szövegben. Ez remek gyakorlás, mert egyszerre használjuk a string bejárást, a feltételt, a számlálót és a `return`-t.

````python
def szamold_meg_a_maganhangzokat(kifejezes: str) -> int:
    """Megszámolja a magánhangzókat egy szövegben (ékezetes betűkkel együtt)."""
    maganhangzok = "aáeéiíoóöőuúüű"
    szamlalo = 0

    for betu in kifejezes.lower():
        if betu in maganhangzok:
            szamlalo += 1

    return szamlalo
````

````python
# Próbáljuk ki különböző szövegekkel!
print(szamold_meg_a_maganhangzokat("Géza kék az ég!"))     # e, ó, i, á = 4
print(szamold_meg_a_maganhangzokat("PyLadies Budapest")) # a, i, e, u, a, e = 6
print(szamold_meg_a_maganhangzokat("Az AI eszközök segíthetik a Python tanulást, a kód megértését!"))          # ü, e, é = 3
````

**Megjegyzés:** Egy függvényen belül nyugodtan használhatsz ciklust, feltételt, lokális változókat — bármit, amit eddig tanultál. A lényeg, hogy a végén a `return` visszaadja az eredményt. Figyeld meg, hogy a magyar ékezetes magánhangzókat is beletettük a listába!

### 1.5 `ugorj_a_listaban()` – Alapértelmezett paraméterek

Feladat: egy listában **ugrálunk** a megadott pozíciótól kezdve, a megadott lépésközzel. Amelyik elemre rálépünk, az bekerül az eredménylistába. Használjunk típusjelölést (`typehinting`-et)

**Alapértelmezett paraméterértékek** (*default parameter values*): ha nem adsz meg egy adott paraméter szerinti argumentumot a hívásnál, a függvény egy előre beállított értéket használ helyette.

````python
def ugorj_a_listaban(lista: list, kezdo_pozicio: int = 0, lepeskoz: int = 1) -> list:
    """A listában ugrál a megadott lépésközzel, és visszaadja az érintett elemeket."""
    eredmeny = []
    pozicio = kezdo_pozicio

    while pozicio < len(lista):
        eredmeny.append(lista[pozicio])
        pozicio += lepeskoz

    return eredmeny
````

````python
gyumolcsok = ["citrom", "banán", "cseresznye", "málna" "eper", "alma", "görögdinnye"]

print("Alapértelmezett értékekkel: 0. pozíciótól, 1-es lépésközzel -> minden elem")
print(ugorj_a_listaban(gyumolcsok))

print("\nCsak a lépésközt adjuk meg: minden 2. elem")
print(ugorj_a_listaban(gyumolcsok, lepeskoz=2))

print("\nA 3. pozíciótól indulva, 2-es lépésközzel")
print(ugorj_a_listaban(gyumolcsok, kezdo_pozicio=3, lepeskoz=2))

print("\nPozicionális argumentumokkal is működik")
print(ugorj_a_listaban(gyumolcsok, 1, 3))

print("\nEz is működik, nevesített argumentumokkal:")
print(ugorj_a_listaban(kezdo_pozicio=1, lista=gyumolcsok))
````

**Tanulság:** Az alapértelmezett paraméterek szuperek, mert a függvényed **rugalmas** lesz: a legegyszerűbb esetben elég csak a listát megadni, de ha finomhangolni akarod, ott vannak a további lehetőségek. Figyeld meg, hogy a hívásnál **kulcsszavas argumentumokkal** (`lepes=2`) is megadhatod az értékeket — ez olvashatóbbá teszi a kódot.

### 1.6 Típusjelölés: hasznos, de nem kötelező

Az előző példákban láttad, hogy a paraméterek és a visszatérési érték mellé **típusjelölést** (*type hint*) írtunk. Fontos tudnod: ez Pythonban **teljesen opcionális** — a Python futtatórendszer (*interpreter*) **nem kényszeríti ki**. A kódod ugyanúgy lefut típusjelöléssel és anélkül is.

Nézzük meg az előző `ugorj_a_listaban()` függvényt típusjelölés **nélkül**:

````python
# Típusjelölés NÉLKÜL — ugyanaz a logika
def lepkedj_a_listaban(lista, kezdo_pozicio=0, lepes=1):
    """A listában ugrál a megadott lépésközzel, és visszaadja az érintett elemeket."""
    eredmeny = []
    pozicio = kezdo_pozicio

    while pozicio < len(lista):
        eredmeny.append(lista[pozicio])
        pozicio += lepes

    return eredmeny
````

````python
# Ugyanúgy működik!
szinek = ["piros", "kék", "zöld", "sárga", "lila", "narancssárga"]
print(lepkedj_a_listaban(szinek, lepes=2))
````

**Hasonlítsuk össze a két verziót:**

```python
# Típusjelöléssel:
def ugorj_a_listaban(lista: list, kezdo_pozicio: int = 0, lepes: int = 1) -> list:

# Típusjelölés nélkül:
def lepkedj_a_listaban(lista, kezdo_pozicio=0, lepes=1):
```

Mindkettő pontosan ugyanazt csinálja! Akkor mégis miért éri meg a típusjelölés?

| Előny | Magyarázat |
|-------|------------|
| **Olvashatóság** | Egy pillantás alatt látod, milyen típusú adatot vár a függvény és mit ad vissza. |
| **Dokumentáció** | A típusjelölés önmagában is dokumentálja a kódot — kevesebb kommentre van szükség. |
| **IDE támogatás** | Az olyan szerkesztők, mint a VS Code, okosabb javaslatokat adnak, ha látják a típusokat. |
| **Hibamegelőzés** | Eszközökkel (pl. `mypy`) ellenőrizheted, hogy nem adsz-e rossz típusú argumentumot. |

**Tanulság:** A típusjelölés egy **ajándék a jövőbeli önmagadnak** (és a csapattársaidnak): segít megérteni a kódot. De ne stresszelj miatta — ha kezdőként elhagyod, a kódod ettől még tökéletesen működik.

## 2. Változók hatóköre (scope) és a LEGB szabály

A **hatókör** (*scope*) határozza meg, hogy egy változó a program melyik részéből érhető el. Egy függvényen belül létrehozott változó alapértelmezetten kívülről nem elérhető — ennek megértése kulcsfontosságú a hibamentes kódhoz.

### 2.1 Bevezető probléma: miért nem működik?

Az alábbi két kód **hibás**. Futtasd le mindkettőt, és figyeld meg, milyen hibát kapsz.

#### Első hibás példa: az egyik függvény nem látja a másik változóit

````python
def inicializal():
    x1 = 1
    x2 = 3
    print(f"inicializal() beállította: x1={x1}, x2={x2}")


def osszead():
    print(f"osszead() megpróbálja összeadni: x1 + x2 = {x1 + x2}")


# Meghívjuk mindkettőt:
inicializal()

try:
    osszead()
except Exception as hiba:
    print(f"HIBA: {hiba}")
````

Az `osszead()` `NameError`-t dobott, annak ellenére, hogy az `inicializal()` korábban létrehozta az `x1` és `x2` változókat.

#### Második hibás példa: a függvény lokális változója kívülről nem elérhető

````python
def lista_osszegzes(lista):
    osszeg = 0
    for elem in lista:
        osszeg += elem
    # Nincs return!
    print(f"A függvényen BELÜL osszeg = {osszeg}")


szamok = [10, 20, 30]
lista_osszegzes(szamok)

try:
    print(f"A függvényen KÍVÜL osszeg = {osszeg}")
except NameError as hiba:
    print(f"HIBA: {hiba}")
````

Az `osszeg` a függvényen belül 60-as értéket kapott, de kívülről nem elérhető — a következő szakasz ezt magyarázza meg.

### 2.2 Magyarázat: mi az a hatókör?

A **hatókör** (*scope*) azt jelöli, hogy a program melyik részéből érhető el egy változó.

A fenti két példa azért nem működött, mert:
1. Az `x1`, `x2` változók az `inicializal()` függvényen belül jöttek létre — az `osszead()` függvényből nem láthatók.
2. Az `osszeg` a `lista_osszegzes()` függvényen belül keletkezett — a külső kód nem fér hozzá.

**Megoldás:** az adatokat **paraméterként** kell átadni, vagy **`return`-nel** visszaadni:

````python
# Javított verzió: return és paraméterek használata
def inicializal():
    x1 = 1
    x2 = 3
    return x1, x2


def osszead(a, b):
    print(f"Összeg: {a} + {b} = {a + b}")


x1, x2 = inicializal()
osszead(x1, x2)
````

Az `inicializal()` `return`-nel adja vissza az értékeket, az `osszead()` pedig paraméterként kapja meg azokat.

### 2.3 A LEGB szabály – Hol keresi a Python a változókat?

A Python egy név (változó, függvény) keresésekor **négy hatókört** vizsgál, ebben a sorrendben:

| Betű | Hatókör | Angol név | Mit jelent? |
|------|---------|-----------|-------------|
| **L** | Lokális | Local | A függvényen belül létrehozott változók |
| **E** | Befoglaló | Enclosing | A külső (befoglaló) függvény változói |
| **G** | Globális | Global | A modul (fájl / notebook) szintjén létrehozott változók |
| **B** | Beépített | Built-in | Python beépített nevei (`print`, `len`, `True`…) |

A keresés a legbelső hatókörből indul és kifelé halad. Ha megtalálta a nevet, megáll. Ha sehol nem találja → `NameError`.

Az alábbiakban mind a négy hatókör egy-egy példával szerepel.

#### **L** – Lokális hatókör (*Local scope*)

A függvényen belül létrehozott változó **lokális** — csak a függvényen belül él, annak befejezésekor megszűnik.

````python
def kedvenc_film_2025():
    # Ez egy LOKÁLIS változó — csak ebben a függvényben létezik
    film = "Pluribus (https://hu.wikipedia.org/wiki/Pluribus_(telev%C3%ADzi%C3%B3s_sorozat))"
    print(f"A kedvenc filmem 2025-ből: {film}")


kedvenc_film_2025()

try:
    print(film)  # Ez hibát dob!
except NameError as hiba:
    print(f"Kívülről nem érhető el: {hiba}")
````

A `film` változó **lokális** (L) — a `kedvenc_film_2025()` függvényen belül jött létre, annak befejeztével megszűnik.

#### **E** – Befoglaló hatókör (*Enclosing scope*)

Ha egy függvényen belül van egy **másik függvény** (belső függvény), a belső függvény eléri a külső függvény változóit. Ez a **befoglaló hatókör**.

````python
def naprendszer():
    # Ez a KÜLSŐ (befoglaló) függvény lokális változója
    bolygok_szama = 8

    def melyik_a_legnagyobb():
        # A BELSŐ függvény "látja" a külső változóját (enclosing scope)
        legnagyobb = "Jupiter"
        print(f"A Naprendszerben {bolygok_szama} bolygó van, a legnagyobb a {legnagyobb}.")

    melyik_a_legnagyobb()


naprendszer()
# melyik_a_legnagyobb() NameError hibát dob
````

A `melyik_a_legnagyobb()` belső függvény hozzáfér a `bolygok_szama` változóhoz, amely a külső `naprendszer()` függvényben lett létrehozva. A LEGB szabály szerint a Python először a lokális (L) hatókörben keresi — nem találja —, majd a befoglaló (E) hatókörben megtalálja.

#### **G** – Globális hatókör (*Global scope*)

A notebook (vagy fájl) szintjén létrehozott változók **globálisak** — függvényekből **olvashatók**, de módosításukhoz a `global` kulcsszó szükséges.

````python
# Ez egy GLOBÁLIS változó — a notebook szintjén jön létre
kedvenc_eger = "Logitech MX Master"


def mutasd_az_egeret():
    # A függvény OLVASNI tudja a globális változót
    print(f"A kedvenc egerem: {kedvenc_eger}")


def probald_megvaltoztatni():
    # Ez NEM a globális változót módosítja!
    # Ehelyett egy ÚJ lokálisat hoz létre ugyanazzal a névvel.
    kedvenc_eger = "Acer AMR 100 Bubble Wireless Mouse"
    print(f"A függvényen belül: {kedvenc_eger}")


mutasd_az_egeret()
probald_megvaltoztatni()
print(f"A globális változó továbbra is: {kedvenc_eger}")

# a kedvenc_eger változót módosítása függvényen belülről a GLOBAL kulcsszó használatával
def nem_adom_fel():
    global kedvenc_eger
    kedvenc_eger = "Logitech MX Anywhere 3S"

nem_adom_fel()
print(f"A globális változó értéke: {kedvenc_eger}")
````

A `probald_megvaltoztatni()` függvényben a `kedvenc_eger = "Acer AMR 100 Bubble Wireless Mouse"` egy **új lokális változót** hoz létre — nem a globálisat módosítja. A globális változó változatlan marad.

> **Megjegyzés:** A globális változó függvényen belüli módosításához a `global` kulcsszó szükséges, de ennek használata általában **kerülendő**, mert nehezen követhető kódot eredményez. Helyette használj `return`-t.

#### **B** – Beépített hatókör (*Built-in scope*)

A Python beépített nevei (`print`, `len`, `int`, `True`…) a **legkülső** hatókörben élnek. Bárhonnan elérhetők, de **felülírhatók** — ami hibákhoz vezethet.

````python
# A print a BEÉPÍTETT hatókörből jön — ezért működik bárhonnan
print("Ez a beépített print() függvény.")
print(f"A print típusa: {type(print)}")
````

````python
# VESZÉLY: felülírjuk a print nevet egy stringgel!
print = "Ez most egy szöveg, nem függvény!"

try:
    print("Hello?")  # Ez hibát dob, mert a print most egy string!
except TypeError as hiba:
    # Nem tudjuk használni a print()-et, mert felülírtuk!
    # De az eredeti megvan a builtins modulban:
    import builtins
    builtins.print(f"HIBA: {hiba}")
    builtins.print("A 'print' nevet felülírtuk egy stringgel!")
````

````python
# Állítsuk helyre a print-et!
del print
print("A print() újra működik!")
````

A `print = "..."` értékadás egy **globális** `print` nevű változót hozott létre. A LEGB szabály szerint a Python a globális (G) hatókörben megtalálta ezt a stringet, és nem jutott el a beépített (B) hatókörig, ahol az eredeti `print()` függvény van.

A `del print` törli a globális `print` változót, így a Python újra a beépített hatókörben találja meg az eredeti `print()` függvényt.

> **Fontos:** Soha ne használj beépített nevet (`print`, `list`, `type`, `sum`, `input` stb.) változónévként — ez nehezen felderíthető hibákat okozhat.
